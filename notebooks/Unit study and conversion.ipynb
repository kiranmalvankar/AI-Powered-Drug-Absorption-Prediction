```python
df['standard_units'] = (
    df['standard_units']
    .astype(str)
    .str.lower()
    .str.strip()
)
```

```python
unit_map = {
    "10'-6 cm/s": "1e-6 cm/s",
    "10^-6 cm/s": "1e-6 cm/s",
    "10e-6 cm/s": "1e-6 cm/s",
    "10e-6 cm s-1": "1e-6 cm/s",
    "1e-6 cm/s": "1e-6 cm/s",
    "10-6 cm/s": "1e-6 cm/s"
}
```

```python
df['unit_normalized'] = df['standard_units'].replace(unit_map)
```
